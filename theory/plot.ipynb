```python
import pandas as pd, matplotlib
matplotlib.rcParams['figure.figsize'] = [20, 9]
```

## Hip Joint

### Square Wave Velocity Controller

```python
df_square = pd.read_csv("../build/square_wave-hip.csv")
df_square["dq"].plot(legend=True)
df_square["dq_r"].plot(legend=True)
df_square["signal"].plot(legend=True, alpha=0.2)
```

```python
df_square["q"].plot()
```

```python
df_sine["q"].plot()
```
